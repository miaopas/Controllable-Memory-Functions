```python
import numpy as np
import matplotlib.pyplot as plt

# Generate example data
length = 50
s_values = np.linspace(0, 1, length)  # Example s-values
y_values = np.zeros(length)

y_values  = np.exp(-1.5*s_values) + np.exp(-2.5*s_values)

y_values  = np.exp(-1.1*s_values) + np.exp(-2*s_values)


n_terms = 2

X = np.column_stack([np.exp(- (n+1) * s_values) for n in range(n_terms)])


a_n, e, _, _ = np.linalg.lstsq(X, y_values, rcond=None)
s_fit = np.linspace(min(s_values), max(s_values), 100)  # Fine grid for smooth plot
terms = n_terms
X_fit = np.column_stack([np.exp(- (n+1) * s_fit) for n in range(terms)])
y_fit = X_fit @ a_n[:terms]  # Compute fitted values


plt.scatter(s_values, y_values, color='red', label="Original Data")
plt.plot(s_fit, y_fit, label="Fitted Curve", linestyle='--', color='blue')
plt.xlabel("s values")
plt.ylabel("y values")
plt.title("Linear Basis Regression: yhat = sum_n an * exp(-n s)")
plt.legend()
plt.grid()
plt.show()

a_n, e
```
