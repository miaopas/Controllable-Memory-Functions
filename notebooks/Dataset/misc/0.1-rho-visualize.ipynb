# impulse

```python
def plot_rho(rho_name, T, num, rho_name2=None):

    # Use the "coolwarm" colormap for better distinction and aesthetics
    cmap = plt.get_cmap(
        "coolwarm"
    )  # Use "coolwarm" or you can try "plasma", "inferno", "viridis", etc.

    cmap2 = plt.get_cmap("binary")
    # Normalize alpha to map the values to colors
    norm = plt.Normalize(0, 1)

    # Adjust the font size for a two-column LaTeX document
    font_size = 22
    plt.rcParams.update(
        {
            "font.size": font_size,  # Set default font size (you can increase/decrease as needed)
            "axes.titlesize": font_size,  # Font size for axis titles
            "axes.labelsize": font_size,  # Font size for axis labels
            "xtick.labelsize": font_size,  # Font size for x-tick labels
            "ytick.labelsize": font_size,  # Font size for y-tick labels
            "legend.fontsize": font_size,  # Font size for legend (if used)
            "figure.titlesize": font_size,  # Font size for the figure title (if applicable)
            # 'text.usetex': True,  # Use LaTeX for rendering text
            # 'font.family': 'serif',  # Set font family to serif (common in LaTeX)
            # 'text.latex.preamble': r'\usepackage{amsmath}'
        }
    )

    # Create the plot
    fig, ax = plt.subplots(figsize=(10, 6))  # Adjust figure size to fit a two-column layout

    # Loop over the range of alpha values
    for i, a in enumerate(np.linspace(0, 1, num)):
        # Get dataset using the LF function (example)
        _, rho = LF(
            data_dir="",
            N=10,
            T=T,
            D=1,
            rho_name=rho_name,
            rho_param={"alpha": a},
            dt=1,
            input_generator="randn",
        ).dataset_construction()

        # Choose the color based on the normalized alpha value
        color = cmap(norm(a))

        # Plot the curve with alpha values on the x-axis (from 0 to 1)
        ax.plot(rho.numpy()[0], color=color)

        if rho_name2 is not None:
            _, rho = LF(
                data_dir="",
                N=10,
                T=T,
                D=1,
                rho_name=rho_name2,
                rho_param={"alpha": a},
                dt=1,
                input_generator="randn",
            ).dataset_construction()

            # Choose the color based on the normalized alpha value
            color = cmap2(norm(a))

            # Plot the curve with alpha values on the x-axis (from 0 to 1)
            ax.plot(rho.numpy()[0], color=color)

    # Set x-axis to show alpha values
    if rho_name2 is not None:
        cbar_ax2 = fig.add_axes(
            [0.812, 0.12, 0.02, 0.8]
        )  # Position for the second colorbar, close to the first
        sm2 = plt.cm.ScalarMappable(cmap=cmap2, norm=norm)
        sm2.set_array([])  # Empty array for colorbar
        cbar2 = plt.colorbar(sm2, cax=cbar_ax2, ticks=[])  # No numerical values
        # cbar2.ax.set_ylabel("Secondary Colormap", rotation=270, labelpad=20)

    cbar_ax1 = fig.add_axes([0.83, 0.12, 0.02, 0.8])  # Position for the first colorbar
    sm1 = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
    sm1.set_array([])  # Empty array for colorbar
    cbar1 = plt.colorbar(sm1, cax=cbar_ax1)
    # cbar1.set_label("Alpha Scale")

    plt.tight_layout(rect=[0, 0, 0.83, 1])  # Adjust layout to make space for the colorbars

    plt.savefig(f"memory-plot-{rho_name}.pdf", bbox_inches="tight")  # Save as PDF for LaTeX

    plt.show()
```

```python
plot_rho("impulse", 512, 6)
```

```python
plot_rho("exponential", 512, 8)
```

```python
plot_rho("polynomial", 512, 6, "exponential")
```
