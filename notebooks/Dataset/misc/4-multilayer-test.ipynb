```python
import numpy as np
from mpmath import mp

# Set the precision to a higher value (e.g., 50)
mp.dps = 50

w = np.abs(np.random.randn(100))


# Define the model function
def model_func(x, *params):
    y = np.zeros_like(x)
    for i, param in enumerate(params):
        y += param * np.exp(-(w[i]) * x)
    return y


# Generate some sample data
np.random.seed(0)
x_data = np.linspace(0, 10, 1000)
y_data = np.exp(-((x_data - 3) ** 2) * 5)


# Construct the design matrix
def construct_design_matrix(x, n_terms):
    X = np.zeros((len(x), n_terms))
    for i in range(n_terms):
        X[:, i] = np.exp(-(i) * x)
    return X


# Construct the design matrix
X = construct_design_matrix(
    x_data, n_terms=100
)  # Adjust n_terms based on the number of coefficients

# Perform the least squares fitting
coefficients, _, _, _ = np.linalg.lstsq(X, y_data, rcond=None)

# Print the fitted coefficients
print("Fitted coefficients:", coefficients)

# Calculate the fitted values
y_fit = model_func(x_data, *coefficients)

# Plot the original data and the fitted curve
import matplotlib.pyplot as plt

plt.plot(x_data, y_data, "bo", label="Original Data")
plt.plot(x_data, y_fit, "r-", label="Fitted Curve")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
G = X.T @ X
eigenvalues = np.linalg.eigvals(G)
np.abs(np.max(eigenvalues) / np.min(eigenvalues))
```
