```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import math
import autoroot
from src.utils.result_processing import MODEL_KEY_REGISTRY


from plot_config import set_plot_style

font_size = 50
legend_font_size = 40
lw=4
figsize = (12,8)

# Apply consistent styling
set_plot_style(font_family='Times New Roman', font_size=font_size, legend_font_size=legend_font_size, figsize=figsize, palette='Spectral')
```

```python
m_key = MODEL_KEY_REGISTRY['LSTM']
df = pd.read_pickle(f"data/LSTM_exp_step3_data_clean.pkl")
df['m'] = df[m_key]
df['alpha'] = df['data.rho_param.alpha']/df['data.rho_param.alpha'].max()
df_min = df
df_min = df_min[df_min['m']==df_min['m'].max()]

fig, ax1 = plt.subplots(figsize=figsize)

# Plot the original lines
palette = sns.color_palette('tab20', n_colors=df_min['run_specs.seed'].nunique())
sns.scatterplot(data=df_min, x='alpha', y='train/rLoss1_epoch.min', ax=ax1, hue='run_specs.seed',
             palette=palette, legend=False)



ax1.set_xlabel(r'$\alpha$')
# ax1.xaxis.set_label_coords(-0.08, 0.03)
ax1.set_yscale('log')
ax1.set_ylabel('Loss')
# ax1.legend(title=r'$m$')

plt.savefig('figs/LSTM_op_adamW_1.pdf', bbox_inches='tight')
```
