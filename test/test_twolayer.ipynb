```python
import numpy as np

m = 2
eigs = [ 0.2+0.1j,0.5+0.1j,0.7+0.1j]
Y = np.diag(eigs)
W = np.array([[1],[1],[1]])
V = np.array([1,1,1])
x = np.array([1,2,3,1,1,1])[:,None]

# Initial f0
f = np.zeros((W.shape[0],))

# List to store all f_t
f_list = [V@f]

for xt in x:
    f = Y @ f + W @ xt
    y = V @ f
    f_list.append(y)

f_list
```

```python
for t in range(x.shape[0]):
    result = 0
    for s in range(t+1):
        result += V@np.linalg.matrix_power(Y, s)@W*x[t-s]
    print(result)
```

```python
alpha = [eigs[i] for i in range(m)]
beta = [eigs[m+i] for i in range(0, m-1)]

eigs

# alpha, beta
```

```python
A2 = np.diag([0]+beta)
A1 = np.diag(alpha)
A1, A2
```

```python
Z = np.array([V[i]*W[i,0] for i in range(2*m-1)])
Zmax = Z.max()
Z
```

```python
B1 = Zmax**(1/3)*np.ones(m)
B1 = B1[:,None]
```

```python
C = [Zmax**(1/3)] + [Z[i]**(1/3) for i in range(m, 2*m-1)]
C
```

```python
B2 = np.zeros((m,m))+0j
for i in range(1,m):
    B2[i,i-1] = (beta[i-1]-alpha[i-1])/(beta[i-1])*(Z[m+i-1]**(2/3))/(Zmax**(1/3))

for j in range(0,m-1):
    B2[0,j] =   (alpha[j])/(beta[j])*(Z[m+j])/(Zmax**(2/3)) +  Z[j]/(Zmax**(2/3)) 

B2[0,m-1] = Z[m-1]/(Zmax**(2/3)) 
B1,B2
```

```python
g = np.zeros((m))
h = np.zeros((m))

# List to store all f_t
# f_list = [V@f]
f_list = []
for xt in x:
    g = A1 @ g + B1 @ xt
    h = A2 @ h + B2 @ g
    y = C @ h
    f_list.append(y)

f_list
```

```python
for t in range(x.shape[0]):
    result = 0
    for s in range(t+1):
        result1 = 0
        for r in range(s+1):
            result += C@np.linalg.matrix_power(A2, r)@B2@np.linalg.matrix_power(A1, s-r)@B1*x[t-s]
        result += result1
    print(result)
```
